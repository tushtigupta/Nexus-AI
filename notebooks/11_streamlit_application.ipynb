```python

from pathlib import Path

project_root = Path("..").resolve()
app_path = project_root / "app.py"

app_code = r"""
import sys
from pathlib import Path

import streamlit as st

PROJECT_ROOT = Path(__file__).resolve().parent
SRC_DIR = PROJECT_ROOT / "src"
MODEL_NAME = "qwen2.5:3b"

if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from multi_agent_coordinator import MultiAgentCoordinator

st.set_page_config(
    page_title="AI Software Engineering Agent",
    page_icon="🤖",
    layout="wide"
)

@st.cache_resource(show_spinner="Loading AI coordinator...")
def load_coordinator():
    return MultiAgentCoordinator(
        project_root=str(PROJECT_ROOT),
        model_name=MODEL_NAME
    )

st.title("🤖 AI Software Engineering Agent")

st.markdown(
    "Analyze Python repositories using AI-powered code retrieval, "
    "code review, security analysis, and automated testing."
)

st.sidebar.header("Analysis Configuration")

repositories = [
    "All repositories", "black", "click", "fastapi", "flask",
    "httpx", "pydantic", "pytest", "requests", "rich", "typer"
]

selected_repository = st.sidebar.selectbox(
    "Select repository", repositories
)

top_k = st.sidebar.slider(
    "Number of code results", 1, 10, 5
)

st.sidebar.caption("Local LLM: qwen2.5:3b")

st.subheader("Ask your Code Assistant")

query = st.text_area(
    "Enter your question",
    placeholder="Example: Explain how requests.get delegates to request",
    height=100
)

if st.button("Run Full Multi-Agent Analysis", type="primary"):
    if not query.strip():
        st.warning("Please enter a question.")
    else:
        try:
            coordinator = load_coordinator()

            repository_filter = (
                None if selected_repository == "All repositories"
                else selected_repository
            )

            with st.spinner("Running all four agents..."):
                result = coordinator.run(
                    query=query.strip(),
                    top_k=top_k,
                    repository=repository_filter
                )

            st.session_state["analysis_result"] = result
            st.success("Analysis execution finished.")

        except Exception as exc:
            st.error(f"Application error: {exc}")

result = st.session_state.get("analysis_result")

if result:
    st.divider()

    st.subheader("Overall Analysis Status")
    st.write(result.get("overall_status", "Unknown"))

    agents = result.get("agents", {})

    retrieval_tab, review_tab, security_tab, testing_tab = st.tabs([
        "Code Retrieval",
        "AI Code Review",
        "Security Analysis",
        "Automated Testing"
    ])

    with retrieval_tab:
        retrieval = agents.get("retrieval", {})
        items = retrieval.get("results", [])

        st.write("Status:", retrieval.get("status"))
        st.write("Results:", len(items))

        for index, item in enumerate(items, 1):
            with st.expander(
                f"{index}. {item.get('name', 'Code result')}"
            ):
                st.write("Repository:", item.get("repository"))
                st.write("File:", item.get("file_path"))
                st.write(
                    "Lines:",
                    item.get("line_start"),
                    "-",
                    item.get("line_end")
                )
                st.write("Type:", item.get("chunk_type"))
                st.code(item.get("code", ""), language="python")

    with review_tab:
        review = agents.get("code_review", {})
        st.write("Status:", review.get("status"))
        st.write(
            "Source chunks reviewed:",
            review.get("source_chunks_reviewed", 0)
        )
        st.markdown(review.get("response", "No review available."))

    with security_tab:
        security = agents.get("security", {})

        st.write("Status:", security.get("status"))
        st.metric(
            "Files scanned",
            security.get("files_scanned", 0)
        )
        st.metric(
            "Potential findings",
            security.get("potential_findings_total", 0)
        )

        st.caption(
            "Potential findings require human verification; "
            "they are not confirmed vulnerabilities."
        )

        for index, finding in enumerate(
            security.get("findings", []), 1
        ):
            with st.expander(
                f"{index}. {finding.get('rule_id')} - "
                f"{finding.get('severity')}"
            ):
                st.write("File:", finding.get("file"))
                st.write("Line:", finding.get("line"))
                st.write("Description:", finding.get("description"))
                st.code(finding.get("evidence", ""), language="python")

    with testing_tab:
        testing = agents.get("testing", {})

        st.write("Status:", testing.get("status"))
        st.write("Tests executed now:", testing.get("tests_executed_now"))

        col1, col2, col3 = st.columns(3)
        col1.metric("Passed", testing.get("passed") or 0)
        col2.metric("Failed", testing.get("failed") or 0)
        col3.metric("Errors", testing.get("errors") or 0)

        st.write("Exit code:", testing.get("exit_code"))

        if testing.get("stdout"):
            with st.expander("Test output"):
                st.code(testing["stdout"])

        if testing.get("stderr"):
            with st.expander("Error output"):
                st.code(testing["stderr"])

    with st.expander("Complete Analysis JSON"):
        st.json(result)

st.divider()

st.caption(
    "AI Software Engineering Agent | RAG | FAISS | "
    "Sentence Transformers | Ollama"
)
"""

app_path.write_text(app_code.strip() + "\n", encoding="utf-8")

print("=" * 70)
print("APPLICATION FILE CREATED")
print("=" * 70)
print("Location:", app_path)
print("Exists:", app_path.exists())
print("Size:", app_path.stat().st_size, "bytes")
```

```python

from pathlib import Path
import py_compile

# ============================================================
# AI SOFTWARE ENGINEERING PLATFORM
# COMPLETE DUAL-MODE STREAMLIT APPLICATION
# ============================================================

project_root = Path("..").resolve()
app_path = project_root / "app.py"

app_code = r'''
import streamlit as st
import sys
import json
import traceback
from pathlib import Path

# ============================================================
# 1. PAGE CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="AI Software Engineering Platform",
    page_icon="🤖",
    layout="wide",
    initial_sidebar_state="expanded"
)

PROJECT_ROOT = Path(__file__).resolve().parent
SRC_PATH = PROJECT_ROOT / "src"

if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

# ============================================================
# 2. CUSTOM STYLING
# ============================================================

st.markdown("""
<style>
[data-testid="stSidebar"] {
    min-width: 290px;
    width: 290px;
}

.main-heading {
    font-size: 34px;
    font-weight: 750;
}

.sidebar-heading {
    font-size: 25px;
    font-weight: 800;
    text-align: center;
    padding: 12px 4px;
}

.stTabs [data-baseweb="tab"] {
    font-weight: 650;
}
</style>
""", unsafe_allow_html=True)

# ============================================================
# 3. LOAD MULTI-AGENT COORDINATOR
# ============================================================

@st.cache_resource
def load_coordinator():

    from multi_agent_coordinator import MultiAgentCoordinator

    return MultiAgentCoordinator(
        project_root=str(PROJECT_ROOT),
        model_name="qwen2.5:3b"
    )

# ============================================================
# 4. SESSION STATE
# ============================================================

if "general_messages" not in st.session_state:
    st.session_state.general_messages = []

if "analysis_result" not in st.session_state:
    st.session_state.analysis_result = None

# ============================================================
# 5. SIDEBAR - APPLICATION SELECTION
# ============================================================

st.sidebar.markdown(
    '<div class="sidebar-heading">🚀 AI PLATFORM</div>',
    unsafe_allow_html=True
)

st.sidebar.markdown("---")

st.sidebar.markdown("### 📌 SELECT APPLICATION MODE")

mode = st.sidebar.radio(
    "Choose your application:",
    [
        "🤖 General AI Assistant",
        "💻 Software Engineering Agent"
    ],
    index=0,
    key="application_mode"
)

st.sidebar.markdown("---")

if mode == "🤖 General AI Assistant":

    st.sidebar.success("🤖 GENERAL AI ASSISTANT")

    st.sidebar.caption(
        "Ask questions, learn programming, understand "
        "technology and explore general knowledge."
    )

    st.sidebar.markdown("**Topics you can explore:**")

    st.sidebar.markdown(
        "- Python\n"
        "- Artificial Intelligence\n"
        "- Machine Learning\n"
        "- Data Science\n"
        "- SQL and databases\n"
        "- Programming concepts"
    )

else:

    st.sidebar.info("💻 SOFTWARE ENGINEERING AGENT")

    st.sidebar.caption(
        "Analyze source code using a retrieval-grounded "
        "multi-agent architecture."
    )

    st.sidebar.markdown("**Available agents:**")

    st.sidebar.markdown(
        "- 🔍 Retrieval Agent\n"
        "- 📝 Code Review Agent\n"
        "- 🛡️ Security Agent\n"
        "- 🧪 Testing Agent"
    )

st.sidebar.markdown("---")
st.sidebar.caption("Powered by Ollama | Qwen 2.5:3b")

# ============================================================
# 6. GENERAL AI ASSISTANT
# ============================================================

if mode == "🤖 General AI Assistant":

    st.markdown(
        '<div class="main-heading">'
        '🤖 General AI Assistant'
        '</div>',
        unsafe_allow_html=True
    )

    st.write(
        "Your personal AI assistant for learning, "
        "programming and everyday questions."
    )

    st.info(
        "Ask anything in simple language. "
        "Get beginner-friendly explanations and examples."
    )

    col1, col2 = st.columns([5, 1])

    with col2:

        if st.button("🗑️ Clear Chat", use_container_width=True):

            st.session_state.general_messages = []
            st.rerun()

    # Display previous conversation

    for message in st.session_state.general_messages:

        with st.chat_message(message["role"]):
            st.markdown(message["content"])

    user_prompt = st.chat_input("Ask me anything...")

    if user_prompt:

        st.session_state.general_messages.append({
            "role": "user",
            "content": user_prompt
        })

        with st.chat_message("user"):
            st.markdown(user_prompt)

        with st.chat_message("assistant"):

            try:

                with st.spinner("Thinking..."):

                    recent_messages = (
                        st.session_state.general_messages[-10:]
                    )

                    conversation = ""

                    for message in recent_messages:

                        conversation += (
                            f"{message['role'].upper()}: "
                            f"{message['content']}\n"
                        )

                    prompt = f"""
You are a helpful general AI assistant.

Explain concepts in beginner-friendly language.
Use practical examples.
For programming questions, provide working code.
Explain code step by step when useful.
Maintain the conversation context.
Do not invent facts.

Conversation:
{conversation}

Assistant:
"""

                    coordinator = load_coordinator()

                    answer = coordinator._ask_llm(prompt)

                    if not answer:
                        answer = "I could not generate an answer. Please try again."

                    st.markdown(answer)

                st.session_state.general_messages.append({
                    "role": "assistant",
                    "content": answer
                })

            except Exception as error:

                st.error(f"Assistant error: {error}")

    if not st.session_state.general_messages:

        st.markdown("### Try asking me:")

        st.markdown(
            "- What is Python?\n"
            "- Explain Artificial Intelligence.\n"
            "- What is the difference between AI and ML?\n"
            "- Write a Python program to add two numbers."
        )

# ============================================================
# 7. SOFTWARE ENGINEERING AGENT
# ============================================================

elif mode == "💻 Software Engineering Agent":

    st.markdown(
        '<div class="main-heading">'
        '💻 Software Engineering Agent'
        '</div>',
        unsafe_allow_html=True
    )

    st.write(
        "An intelligent multi-agent system for code retrieval, "
        "code review, security analysis and testing."
    )

    st.success(
        "Four integrated agents: Retrieval, Code Review, "
        "Security and Testing."
    )

    # --------------------------------------------------------
    # Repository selection
    # --------------------------------------------------------

    repositories = [
        "black",
        "click",
        "fastapi",
        "flask",
        "httpx",
        "pydantic",
        "pytest",
        "requests",
        "rich",
        "typer"
    ]

    col1, col2 = st.columns(2)

    with col1:

        repository = st.selectbox(
            "Select Repository",
            repositories,
            index=repositories.index("requests")
        )

    with col2:

        top_k = st.slider(
            "Number of Retrieved Code Chunks",
            1,
            10,
            5
        )

    # --------------------------------------------------------
    # User query
    # --------------------------------------------------------

    query = st.text_area(
        "Enter your software engineering question:",
        placeholder=(
            "Example: Explain how requests.get() "
            "delegates to request()."
        ),
        height=110
    )

    run_analysis = st.button(
        "🚀 Run Full Multi-Agent Analysis",
        type="primary",
        use_container_width=True
    )

    # --------------------------------------------------------
    # Run all four agents
    # --------------------------------------------------------

    if run_analysis:

        if not query.strip():

            st.warning("Please enter a question first.")

        else:

            st.session_state.analysis_result = None

            try:

                with st.spinner(
                    "Running all four AI agents. Please wait..."
                ):

                    coordinator = load_coordinator()

                    result = coordinator.run(
                        query=query,
                        top_k=top_k,
                        repository=repository
                    )

                    st.session_state.analysis_result = result

                if isinstance(result, dict):

                    if result.get("overall_status") == "completed":
                        st.success("Multi-agent analysis completed.")

                    else:
                        st.warning(
                            "Analysis completed with warnings. "
                            "Inspect individual agent results."
                        )

                else:
                    st.warning(
                        "Coordinator returned an unexpected result."
                    )

            except Exception as error:

                st.error(f"Multi-agent analysis failed: {error}")
                st.code(traceback.format_exc())

    # ========================================================
    # 8. DISPLAY AGENT RESULTS
    # ========================================================

    result = st.session_state.analysis_result

    if result is not None:

        st.markdown("---")
        st.markdown("## 📊 Multi-Agent Analysis Results")

        if not isinstance(result, dict):

            st.warning("Unexpected result format.")
            st.write(result)

        else:

            # IMPORTANT:
            # The coordinator stores all four agents under "agents".

            agents = result.get("agents", {})

            if not agents:

                st.error("No agent results found in coordinator output.")

                with st.expander("Inspect raw coordinator response"):
                    st.json(result)

            else:

                # ------------------------------------------------
                # Overall status
                # ------------------------------------------------

                overall_status = result.get(
                    "overall_status",
                    "unknown"
                )

                st.info(
                    f"Overall Status: {overall_status.upper()}"
                )

                st.write(
                    "**Repository:**",
                    result.get("repository_filter", "Not specified")
                )

                st.write(
                    "**Query:**",
                    result.get("query", "")
                )

                # ------------------------------------------------
                # Agent status summary
                # ------------------------------------------------

                st.markdown("### Agent Execution Summary")

                summary_columns = st.columns(4)

                agent_information = [
                    ("retrieval", "🔍 Retrieval"),
                    ("code_review", "📝 Code Review"),
                    ("security", "🛡️ Security"),
                    ("testing", "🧪 Testing")
                ]

                for column, (key, label) in zip(
                    summary_columns,
                    agent_information
                ):

                    agent_data = agents.get(key, {})

                    status = agent_data.get(
                        "status",
                        "missing"
                    )

                    with column:

                        st.metric(
                            label,
                            str(status).upper()
                        )

                st.markdown("---")

                # ------------------------------------------------
                # Four independent tabs
                # ------------------------------------------------

                retrieval_tab, review_tab, security_tab, testing_tab = (
                    st.tabs([
                        "🔍 Retrieval",
                        "📝 Code Review",
                        "🛡️ Security",
                        "🧪 Testing"
                    ])
                )

                # =================================================
                # AGENT 1 - RETRIEVAL
                # =================================================

                with retrieval_tab:

                    st.subheader("🔍 Retrieval Agent")

                    retrieval = agents.get(
                        "retrieval",
                        {}
                    )

                    st.write(
                        "**Status:**",
                        retrieval.get("status", "unknown")
                    )

                    st.metric(
                        "Retrieved Code Chunks",
                        retrieval.get("results_count", 0)
                    )

                    retrieved_results = retrieval.get(
                        "results",
                        []
                    )

                    if retrieved_results:

                        st.markdown("### Retrieved Source Code")

                        for index, item in enumerate(
                            retrieved_results,
                            start=1
                        ):

                            with st.expander(
                                f"Code Chunk {index}"
                            ):

                                st.json(item)

                    else:

                        st.warning(
                            "No code chunks were retrieved."
                        )

                    if retrieval.get("error"):
                        st.error(retrieval["error"])

                # =================================================
                # AGENT 2 - CODE REVIEW
                # =================================================

                with review_tab:

                    st.subheader("📝 Code Review Agent")

                    review = agents.get(
                        "code_review",
                        {}
                    )

                    st.write(
                        "**Status:**",
                        review.get("status", "unknown")
                    )

                    if review.get("error"):
                        st.error(review["error"])

                    # Show all actual fields returned by the agent.
                    # This avoids assuming a particular output schema.

                    review_content = {
                        key: value
                        for key, value in review.items()
                        if key not in {"status", "error"}
                    }

                    if review_content:

                        for key, value in review_content.items():

                            st.markdown(
                                f"### {key.replace('_', ' ').title()}"
                            )

                            if isinstance(value, str):
                                st.markdown(value)

                            elif isinstance(value, (dict, list)):
                                st.json(value)

                            else:
                                st.write(value)

                    elif not review.get("error"):

                        st.warning(
                            "The code review agent returned no "
                            "additional output fields."
                        )

                    with st.expander(
                        "View Complete Code Review Response"
                    ):

                        st.json(review)

                # =================================================
                # AGENT 3 - SECURITY
                # =================================================

                with security_tab:

                    st.subheader("🛡️ Security Agent")

                    security = agents.get(
                        "security",
                        {}
                    )

                    st.write(
                        "**Status:**",
                        security.get("status", "unknown")
                    )

                    if "new_scan_performed" in security:

                        st.write(
                            "**Fresh scan performed:**",
                            security["new_scan_performed"]
                        )

                    if security.get("error"):
                        st.error(security["error"])

                    # Display every field actually returned.

                    security_content = {
                        key: value
                        for key, value in security.items()
                        if key not in {"status", "error"}
                    }

                    if security_content:

                        for key, value in security_content.items():

                            st.markdown(
                                f"### {key.replace('_', ' ').title()}"
                            )

                            if isinstance(value, str):
                                st.markdown(value)

                            elif isinstance(value, (dict, list)):
                                st.json(value)

                            else:
                                st.write(value)

                    else:

                        st.warning(
                            "No additional security output was returned."
                        )

                    with st.expander(
                        "View Complete Security Response"
                    ):

                        st.json(security)

                # =================================================
                # AGENT 4 - TESTING
                # =================================================

                with testing_tab:

                    st.subheader("🧪 Testing Agent")

                    testing = agents.get(
                        "testing",
                        {}
                    )

                    st.write(
                        "**Status:**",
                        testing.get("status", "unknown")
                    )

                    if testing.get("error"):
                        st.error(testing["error"])

                    # Display every actual field returned.

                    testing_content = {
                        key: value
                        for key, value in testing.items()
                        if key not in {"status", "error"}
                    }

                    if testing_content:

                        for key, value in testing_content.items():

                            st.markdown(
                                f"### {key.replace('_', ' ').title()}"
                            )

                            if isinstance(value, str):
                                st.markdown(value)

                            elif isinstance(value, (dict, list)):
                                st.json(value)

                            else:
                                st.write(value)

                    else:

                        st.warning(
                            "No additional testing output was returned."
                        )

                    with st.expander(
                        "View Complete Testing Response"
                    ):

                        st.json(testing)

                # ------------------------------------------------
                # Complete result
                # ------------------------------------------------

                with st.expander(
                    "📄 View Complete Multi-Agent Output"
                ):

                    st.json(result)

                # ------------------------------------------------
                # Limitations
                # ------------------------------------------------

                limitations = result.get(
                    "limitations",
                    []
                )

                if limitations:

                    with st.expander(
                        "⚠️ Analysis Limitations"
                    ):

                        for limitation in limitations:
                            st.markdown(f"- {limitation}")

# ============================================================
# 9. FOOTER
# ============================================================

st.markdown("---")

st.caption(
    "AI Software Engineering Platform | "
    "General AI Assistant + Multi-Agent Software Engineering Agent | "
    "Ollama + Qwen 2.5"
)
'''

# ============================================================
# 10. SAVE AND VALIDATE
# ============================================================

app_path.write_text(
    app_code,
    encoding="utf-8"
)

print("=" * 65)
print("AI SOFTWARE ENGINEERING PLATFORM")
print("=" * 65)

print("Application file:", app_path)
print("File exists:", app_path.exists())
print("File size:", app_path.stat().st_size, "bytes")

py_compile.compile(
    str(app_path),
    doraise=True
)

print("\nPython syntax: VALID")
print("\nApplication modes:")
print("1. General AI Assistant")
print("2. Software Engineering Agent")

print("\nFour integrated result tabs:")
print("1. Retrieval")
print("2. Code Review")
print("3. Security")
print("4. Testing")

print("\nNested coordinator result handling: ENABLED")
print("=" * 65)
```
