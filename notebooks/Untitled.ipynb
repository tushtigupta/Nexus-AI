```python

import ast
import importlib.metadata as metadata
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\Tushti Gupta\AI_Software_Engineering_Agent"
)

files_to_check = [
    PROJECT_ROOT / "app.py"
]

src_folder = PROJECT_ROOT / "src"

if src_folder.exists():
    files_to_check.extend(src_folder.rglob("*.py"))

stdlib_modules = set(getattr(__import__("sys"), "stdlib_module_names", []))

external_imports = set()
failed_files = []

for file in files_to_check:
    try:
        source = file.read_text(encoding="utf-8")
        tree = ast.parse(source)

        for node in ast.walk(tree):
            if isinstance(node, ast.Import):
                for alias in node.names:
                    external_imports.add(alias.name.split(".")[0])

            elif isinstance(node, ast.ImportFrom):
                if node.module:
                    external_imports.add(node.module.split(".")[0])

    except Exception as e:
        failed_files.append((str(file), str(e)))

external_imports -= stdlib_modules

print("=" * 65)
print("NEXUS AI - ACTUAL PROJECT DEPENDENCIES")
print("=" * 65)

print("\nPython files inspected:", len(files_to_check))
print("\nEXTERNAL IMPORTS AND INSTALLED VERSIONS\n")

package_map = metadata.packages_distributions()

for module in sorted(external_imports):
    distributions = package_map.get(module, [])

    if distributions:
        versions = []
        for dist in distributions:
            try:
                versions.append(
                    f"{dist}=={metadata.version(dist)}"
                )
            except metadata.PackageNotFoundError:
                pass

        print(f"{module:25} | {', '.join(versions) or 'Version unavailable'}")
    else:
        print(f"{module:25} | Local module or distribution not mapped")

print("\nFILES WITH INSPECTION ERRORS:", len(failed_files))
for file, error in failed_files[:10]:
    print(file, "|", error)

print("\nINSPECTION COMPLETED")
print("=" * 65)
```

```python

from pathlib import Path

ROOT = Path(r"C:\Users\Tushti Gupta\AI_Software_Engineering_Agent")

FILES = [
    ROOT / "app.py",
    ROOT / "src" / "multi_agent_coordinator.py"
]

keywords = [
    "code_chunks",
    "faiss_index",
    "vector_metadata",
    "embedding_metadata",
    "raw_repositories",
    "extracted_code",
    "phase8_final_report",
    "phase9_static_security_findings",
    "open(",
    "read_text",
    "load_local",
    "np.load",
    "faiss.read_index",
    "Path("
]

print("=" * 75)
print("NEXUS AI - ACTIVE RUNTIME DEPENDENCY INSPECTION")
print("=" * 75)

for file in FILES:
    print(f"\n\nFILE: {file.relative_to(ROOT)}")
    print("-" * 75)

    if not file.exists():
        print("FILE NOT FOUND")
        continue

    lines = file.read_text(
        encoding="utf-8",
        errors="ignore"
    ).splitlines()

    matching_lines = []

    for i, line in enumerate(lines):
        if any(word.lower() in line.lower() for word in keywords):
            start = max(0, i - 2)
            end = min(len(lines), i + 3)

            matching_lines.append((start, end))

    # Merge overlapping context ranges
    merged = []

    for start, end in matching_lines:
        if merged and start <= merged[-1][1]:
            merged[-1] = (
                merged[-1][0],
                max(merged[-1][1], end)
            )
        else:
            merged.append((start, end))

    for start, end in merged:
        print(f"\n--- Lines {start + 1}-{end} ---")
        for j in range(start, end):
            print(f"{j + 1:4}: {lines[j]}")

print("\n" + "=" * 75)
print("INSPECTION COMPLETE - NO FILES MODIFIED")
print("=" * 75)
```
