```python
import os
import shutil
from datetime import datetime

PROJECT_PATH = r"C:\Users\Tushti Gupta\AI_Software_Engineering_Agent"

# Create backup beside the original project
parent_folder = os.path.dirname(PROJECT_PATH)
project_name = os.path.basename(PROJECT_PATH)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
BACKUP_PATH = os.path.join(
    parent_folder,
    f"{project_name}_BACKUP_{timestamp}"
)

# Exclude temporary and unnecessary folders
EXCLUDE_FOLDERS = {
    "__pycache__",
    ".git",
    ".ipynb_checkpoints",
    ".venv",
    "venv",
    "env"
}

def ignore_items(directory, names):
    return [
        name for name in names
        if name in EXCLUDE_FOLDERS
        or name.startswith("BACKUP_")
    ]

print("Creating Phase 11 project backup...")

if not os.path.isdir(PROJECT_PATH):
    print("ERROR: Original project folder not found.")

elif os.path.exists(BACKUP_PATH):
    print("ERROR: Backup destination already exists. No files copied.")

else:
    shutil.copytree(
        PROJECT_PATH,
        BACKUP_PATH,
        ignore=ignore_items
    )

    print("\nBACKUP CREATED SUCCESSFULLY")
    print("Backup location:", BACKUP_PATH)

    # Verify important files in backup
    files_to_check = [
        "src/multi_agent_coordinator.py",
        "results/phase10_live_validation.json",
        "notebooks/test_phase8_generated.py",
        "notebooks/test_phase8_api_wrappers.py",
        "notebooks/test_phase8_corrected_wrappers.py"
    ]

    print("\nBACKUP VERIFICATION:")

    for file in files_to_check:
        backup_file = os.path.join(BACKUP_PATH, file)
        status = "FOUND" if os.path.isfile(backup_file) else "MISSING"
        print(f"{file}: {status}")

    print("\nBackup size (MB):",
          round(sum(
              os.path.getsize(os.path.join(root, f))
              for root, _, filenames in os.walk(BACKUP_PATH)
              for f in filenames
          ) / (1024 * 1024), 2))

print("\nPHASE 11 - STEP 2 FINISHED")
```

```python
import os
import json

PROJECT_PATH = r"C:\Users\Tushti Gupta\AI_Software_Engineering_Agent"
RESULTS_PATH = os.path.join(PROJECT_PATH, "results")

report_files = [
    "phase9_static_security_findings.json",
    "phase10_live_validation.json"
]

for filename in report_files:

    file_path = os.path.join(RESULTS_PATH, filename)

    print("\n" + "=" * 70)
    print("REPORT:", filename)
    print("=" * 70)

    if not os.path.isfile(file_path):
        print("FILE NOT FOUND")
        continue

    with open(file_path, "r", encoding="utf-8") as f:
        report = json.load(f)

    print("Top-level keys:", list(report.keys()))

    # Display security-related information
    def inspect_security(data, path="root", depth=0):
        if depth > 4:
            return

        if isinstance(data, dict):
            for key, value in data.items():
                if any(word in key.lower() for word in
                       ["security", "finding", "vulnerability", "warning"]):
                    print(f"\n[{path}.{key}]")
                    print(json.dumps(value, indent=2, default=str)[:7000])

                if isinstance(value, (dict, list)):
                    inspect_security(value, f"{path}.{key}", depth + 1)

        elif isinstance(data, list):
            for index, item in enumerate(data[:3]):
                if isinstance(item, (dict, list)):
                    inspect_security(item, f"{path}[{index}]", depth + 1)

    inspect_security(report)

print("\nSECURITY REPORT INSPECTION FINISHED")
```

```python

from collections import Counter

print("=" * 75)
print("PHASE 11 - METADATA ALIGNMENT VALIDATION")
print("=" * 75)

# 1. Check record counts
print("\n1. RECORD COUNTS")
print("Code chunks:", len(code_chunks))
print("Vector metadata:", len(vector_metadata))

# 2. Check whether corresponding records match
fields_to_check = [
    "chunk_id",
    "chunk_type",
    "name",
    "line_start",
    "line_end"
]

mismatches = []

for i, (chunk, metadata) in enumerate(
    zip(code_chunks, vector_metadata)
):
    for field in fields_to_check:
        if chunk.get(field) != metadata.get(field):
            mismatches.append({
                "index": i,
                "field": field,
                "chunk_value": chunk.get(field),
                "metadata_value": metadata.get(field)
            })

print("\n2. CHUNK-METADATA ALIGNMENT")
print("Mismatched fields:", len(mismatches))

if mismatches:
    print("\nFirst 10 mismatches:")
    for item in mismatches[:10]:
        print(item)
else:
    print("All checked fields match.")

# 3. Check vector ID ordering
vector_ids = [m.get("vector_id") for m in vector_metadata]

sequential_ids = all(
    vector_id == i
    for i, vector_id in enumerate(vector_ids)
)

print("\n3. VECTOR ID ORDER")
print("Sequential vector IDs:", sequential_ids)

# 4. Repository distribution
repository_counts = Counter(
    m.get("repository", "Unknown")
    for m in vector_metadata
)

print("\n4. REPOSITORY DISTRIBUTION")
for repository, count in sorted(repository_counts.items()):
    print(f"{repository}: {count}")

# 5. Chunk type distribution
chunk_type_counts = Counter(
    m.get("chunk_type", "Unknown")
    for m in vector_metadata
)

print("\n5. CHUNK TYPE DISTRIBUTION")
for chunk_type, count in sorted(chunk_type_counts.items()):
    print(f"{chunk_type}: {count}")

# 6. Inspect a few records from different repositories
print("\n6. SAMPLE METADATA RECORDS")

shown = set()

for metadata in vector_metadata:
    repository = metadata.get("repository")

    if repository not in shown:
        shown.add(repository)
        print({
            "repository": repository,
            "chunk_id": metadata.get("chunk_id"),
            "name": metadata.get("name"),
            "chunk_type": metadata.get("chunk_type"),
            "file_path": metadata.get("file_path")
        })

    if len(shown) >= 5:
        break

print("\nMETADATA ALIGNMENT VALIDATION COMPLETED")
```

```python

import numpy as np

# Manually reviewed relevance labels for the six benchmark queries.
relevance_labels = {
    "Q1": [1, 0, 1, 1, 0],
    "Q2": [1, 1, 1, 0, 0],
    "Q3": [1, 1, 0, 0, 0],
    "Q4": [0, 1, 0, 0, 0],
    "Q5": [0, 1, 1, 1, 0],
    "Q6": [1, 0, 0, 0, 0]
}

# Verify that labels correspond to the six retrieved queries.
assert set(relevance_labels.keys()) == set(retrieval_results.keys())
assert all(len(labels) == 5 for labels in relevance_labels.values())

evaluation_results = []

for query_id, labels in relevance_labels.items():

    # Precision at 5
    precision_at_5 = sum(labels) / 5

    # Precision at 3
    precision_at_3 = sum(labels[:3]) / 3

    # Reciprocal Rank
    first_relevant_rank = next(
        (rank for rank, label in enumerate(labels, start=1) if label == 1),
        None
    )

    reciprocal_rank = (
        1 / first_relevant_rank
        if first_relevant_rank is not None
        else 0
    )

    evaluation_results.append({
        "Query": query_id,
        "Relevant Results": sum(labels),
        "Precision@3": round(precision_at_3, 4),
        "Precision@5": round(precision_at_5, 4),
        "First Relevant Rank": first_relevant_rank,
        "Reciprocal Rank": round(reciprocal_rank, 4)
    })

# Aggregate metrics
mean_precision_at_3 = np.mean(
    [item["Precision@3"] for item in evaluation_results]
)

mean_precision_at_5 = np.mean(
    [item["Precision@5"] for item in evaluation_results]
)

mean_reciprocal_rank = np.mean(
    [item["Reciprocal Rank"] for item in evaluation_results]
)

print("=" * 85)
print("PHASE 11 - RETRIEVAL EVALUATION RESULTS")
print("=" * 85)

for item in evaluation_results:
    print("\n", item)

print("\n" + "=" * 85)
print("OVERALL METRICS")
print("=" * 85)

print("Benchmark queries:", len(evaluation_results))
print("Total retrieved chunks:", sum(len(v) for v in relevance_labels.values()))
print("Total relevant chunks:", sum(sum(v) for v in relevance_labels.values()))
print("Mean Precision@3:", round(mean_precision_at_3, 4))
print("Mean Precision@5:", round(mean_precision_at_5, 4))
print("Mean Reciprocal Rank (MRR):", round(mean_reciprocal_rank, 4))

print("\nRETRIEVAL EVALUATION COMPLETED")
```

```python

import os
import json
from datetime import datetime

# Original benchmark questions
benchmark_queries = {
    "Q1": "How does the Requests library send an HTTP request?",
    "Q2": "How does HTTPX implement its top-level request function?",
    "Q3": "How does Flask register a route handler?",
    "Q4": "How does FastAPI process an incoming request?",
    "Q5": "How does Pydantic validate model data?",
    "Q6": "How does pytest execute a test function?"
}

# Prepare manually judged retrieval records
benchmark_records = []

for query_id, labels in relevance_labels.items():
    retrieved_chunks = []

    for rank, (result, label) in enumerate(
        zip(retrieval_results[query_id], labels),
        start=1
    ):
        retrieved_chunks.append({
            "rank": rank,
            "chunk_id": result.get("chunk_id"),
            "name": result.get("name"),
            "chunk_type": result.get("chunk_type"),
            "repository": result.get("repository"),
            "file_path": result.get("file_path"),
            "relevance": label
        })

    benchmark_records.append({
        "query_id": query_id,
        "query": benchmark_queries[query_id],
        "relevant_chunks": sum(labels),
        "precision_at_3": round(sum(labels[:3]) / 3, 4),
        "precision_at_5": round(sum(labels) / 5, 4),
        "first_relevant_rank": next(
            (i for i, value in enumerate(labels, start=1) if value == 1),
            None
        ),
        "reciprocal_rank": round(
            next(
                (1 / i for i, value in enumerate(labels, start=1) if value == 1),
                0
            ),
            4
        ),
        "retrieved_chunks": retrieved_chunks
    })

# Final evaluation report
phase11_retrieval_report = {
    "phase": 11,
    "title": "Retrieval Evaluation",
    "evaluation_date": datetime.now().isoformat(timespec="seconds"),
    "dataset_size": 30650,
    "benchmark_query_count": len(benchmark_records),
    "total_retrieved_chunks": 30,
    "total_relevant_chunks": 13,
    "metrics": {
        "mean_precision_at_3": round(mean_precision_at_3, 4),
        "mean_precision_at_5": round(mean_precision_at_5, 4),
        "mean_reciprocal_rank": round(mean_reciprocal_rank, 4)
    },
    "relevance_criteria": {
        "1": "Directly relevant to answering the query",
        "0": "Not directly relevant or only partially related"
    },
    "methodology": (
        "Six benchmark queries were evaluated. The top five retrieved "
        "chunks for each query were manually inspected and assigned "
        "binary relevance labels."
    ),
    "limitations": [
        "The benchmark contains only six queries.",
        "Relevance judgments cover only the top five retrieved chunks.",
        "Recall was not calculated because an exhaustive relevant-chunk "
        "ground-truth set was not established.",
        "Results represent an exploratory evaluation, not a complete "
        "measurement of retrieval quality across the entire dataset."
    ],
    "query_results": benchmark_records
}

os.makedirs("results", exist_ok=True)

report_path = os.path.join(
    "results",
    "phase11_retrieval_evaluation.json"
)

with open(report_path, "w", encoding="utf-8") as file:
    json.dump(phase11_retrieval_report, file, indent=4)

print("=" * 85)
print("PHASE 11 - RETRIEVAL REPORT SAVED")
print("=" * 85)
print("Location:", os.path.abspath(report_path))
print("Benchmark queries:", len(benchmark_records))
print("Retrieved chunks:", 30)
print("Relevant chunks:", 13)
print("Precision@3:", mean_precision_at_3)
print("Precision@5:", mean_precision_at_5)
print("MRR:", mean_reciprocal_rank)
print("\nReport exists:", os.path.exists(report_path))
```

```python
import os
import ast
import inspect

project_root = os.path.abspath("..")
coordinator_path = os.path.join(
    project_root,
    "src",
    "multi_agent_coordinator.py"
)

print("=" * 85)
print("PHASE 11 - MULTI-AGENT COORDINATOR INSPECTION")
print("=" * 85)

if not os.path.exists(coordinator_path):
    raise FileNotFoundError(coordinator_path)

with open(coordinator_path, "r", encoding="utf-8") as file:
    source = file.read()

tree = ast.parse(source)

for node in tree.body:
    if isinstance(node, ast.ClassDef):
        print("\nCLASS:", node.name)

        for item in node.body:
            if isinstance(item, (ast.FunctionDef, ast.AsyncFunctionDef)):
                print(
                    f"Method: {item.name} | "
                    f"Line: {item.lineno} | "
                    f"Arguments: {[arg.arg for arg in item.args.args]}"
                )

print("\nCoordinator source:", coordinator_path)
print("Source size:", len(source), "characters")
print("\nCOORDINATOR INSPECTION COMPLETED")
```
